```python
# Filtering dataframe
filtered_df = df[df['Age']>28]
print(filtered_df)
```

```python
# Multiple Condtions 
# And Consitions

filtered_df = df[(df['Age']>28) & (df['City'] =='Bhaktapur')]
print(filtered_df)
```

```python
# OR Consitions

filtered_df = df[(df['Age']>28) | (df['City'] =='Bhaktapur')]
print(filtered_df)
```

```python
# And Consitions

filtered_df = df[ not ((df['Age']>28) & (df['City'] =='Bhaktapur'))]
print(filtered_df)
```

```python
filtered_df = df[ not (df['Age']>28)]
print(filtered_df)
```

```python
filtered_df = df[ ((df['Age']>28) , (df['City'] =='Bhaktapur'))]
print(filtered_df)
```

```python
filtered_df = df[ (df['Age']>28)  (df['City'] =='Bhaktapur')]
print(filtered_df)
```

CRUD -> ADD, Update, Delete (to the existing table)

```python
# ADD a new column
df['Salary'] = [50000, 60000, 70000]
print(df)
```

```python
# Update existing column
df['Salary'] = df['Salary'] + 5000
print(df)
```

```python
# Update existing column with condition
df.loc[df['Age']>30, 'Salary'] = df['Salary'] + 10000
print(df)
```

```python
# Deleting a column
df.drop('Salary', axis=1, inplace=True)
print(df)
```

```python
df.drop('Salary')
print(df)
```

```python
df.drop('Salary', axis=0, inplace=True)
print(df)
```

```python
df.drop(0, axis=1, inplace=True)
print(df)
```

```python
# Delete row
df.drop(0, axis=0, inplace=True)
print(df)
```
