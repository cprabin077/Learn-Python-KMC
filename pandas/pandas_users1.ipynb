```python
# Matplotib visualization
top_states = state_counts.sort_values(by='user_count', ascending=False).head(10)
plt.figure(figsize=(10,6))
plt.bar(top_states['state'], top_states['user_count'], color = 'green')
plt.xlabel('State')
plt.ylabel('Number of states')
plt.title('Top 10 States by User Count')
plt.xticks(rotation = 45)
plt.tight_layout()
plt.show()
```

```python
# Save image file
plt.savefig('top_states_bar_graph.png', dpi=300, bbox_inches='tight')

plt.show()
```

```python
# Matplotlib Pie Chart Visualization
top_states = state_counts.sort_values(by='user_count', ascending=False).head(10)

plt.figure(figsize=(10,8))
plt.pie(
    top_states['user_count'],
    labels=top_states['state'],
    autopct='%1.1f%%',
    startangle=90
)

plt.title('Top 10 States by User Count')
plt.tight_layout()
plt.show()
```

```python
# Matplotlib Line Graph Visualization
top_states = state_counts.sort_values(by='user_count', ascending=False).head(10)

plt.figure(figsize=(10,6))

# Line graph
plt.plot(
    top_states['state'],
    top_states['user_count'],
    marker='o'
)

plt.xlabel('State')
plt.ylabel('Number of Users')
plt.title('Top 10 States by User Count')

# Rotate x-axis labels
plt.xticks(rotation=45)

# Add grid lines
plt.grid(axis='y', linestyle='--', alpha=0.7)

plt.tight_layout()

# Save image
plt.savefig('top_states_line_graph.png', dpi=300, bbox_inches='tight')

plt.show()
```
