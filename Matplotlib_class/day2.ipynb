Use a dotted line:

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
ypoints = np.array([3, 8, 1, 10])

plt.plot(ypoints, linestyle = 'dotted')
plt.show()
```

Use a dashed line:

```python
plt.plot(ypoints, linestyle = 'dashed')
```

Set the line color to red:

```python
plt.plot(ypoints, color = 'r')
plt.show()
```
