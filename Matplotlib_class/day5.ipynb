Display Multiple Plots

Draw 2 plots:

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
#plot 1:
x = np.array([0, 1, 2, 3])
y = np.array([3, 8, 1, 10])

plt.subplot(1, 2, 2) #the figure has 1 row, 2 columns, and this plot is the second plot.
plt.plot(x,y)

#plot 2:
x = np.array([0, 1, 2, 3])
y = np.array([10, 20, 30, 40])

plt.subplot(1, 2, 1) #the figure has 1 row, 2 columns, and this plot is the first plot
plt.plot(x,y)

plt.show()
```

Draw 2 plots on top of each other:


 2 rows and 1 column (meaning that the two plots will be displayed on top of each other instead of side-by-side)

```python
#plot 1:
x = np.array([0, 1, 2, 3])
y = np.array([3, 8, 1, 10])

plt.subplot(2, 1, 1) # 2 rows and 1 column and 1 at last means display in the top
plt.plot(x,y)

#plot 2:
x = np.array([0, 1, 2, 3])
y = np.array([10, 20, 30, 40])

plt.subplot(2, 1, 2) # 2 rows and 1 column display and 2 at last means display in the below of the top
plt.plot(x,y)

plt.show()
```
